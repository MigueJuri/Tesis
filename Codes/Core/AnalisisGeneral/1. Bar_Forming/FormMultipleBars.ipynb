```python
import pandas as pd


def load_data_from_csv(csv_file: str) -> pd.DataFrame:
    # yfinance CSV export uses a 2-row header for MultiIndex columns and a standalone index-name row.
    # data = pd.read_csv(csv_file, header=[0, 1], skiprows=[2], index_col=0)

    # data.index = pd.to_datetime(data.index)
    # data.index.name = "timestamp"
    df = pd.read_csv(csv_file, parse_dates=['timestamp'])
    df.sort_values('timestamp', inplace=True)

    return df
```

```python
import pandas as pd

def load_tickers_from_file(file_path: str) -> list:
    """
    Loads tickers from a text file, one ticker per line.
    """
    with open(file_path, 'r') as f:
        tickers = [line.strip() for line in f if line.strip()]
    return tickers
ewma_window = 252
ticker_file = r"G:\Mi unidad\2026\Tesis-1\Códigos\Data\tickers\sp500_tickers_Energy.csv"

tickers = load_tickers_from_file(ticker_file)

for ticker in tickers:

    
    output_file = rf"G:\Mi unidad\2026\Tesis-1\Códigos\Data\DataTickers\{ticker}_2015_to_2026_1Min.csv"
    df = load_data_from_csv(output_file)
    print(f"Processing {ticker}...")
    
    dollar_bars = fd.form_bars(df, bar_type="dollar", ewma_window=ewma_window)
    dollar_bars.to_csv(rf"G:\Mi unidad\2026\Tesis-1\Códigos\Data\DataTickersDollar\{ticker}_2015_to_2026_dollar_bars.csv")
```
