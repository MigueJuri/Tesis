Simulación de Movimiento Browniano

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_brownian_motion(num_paths=20, num_steps=1000, dt=0.1):
    # Generate time array
    t = np.linspace(0, num_steps * dt, num_steps)
    
    # Efficiently generate multiple Brownian motion paths using vectorization
    # Shape: (num_steps, num_paths)
    increments = np.zeros((num_steps, num_paths))
    increments[1:] = np.random.normal(0, np.sqrt(dt), (num_steps - 1, num_paths))
    paths = np.cumsum(increments, axis=0)
    
    # Plot all paths
    plt.figure(figsize=(12, 6))
    for i in range(num_paths):
        plt.plot(t, paths[:, i], alpha=0.7)

    # # Calculate standard deviation across all paths at each time step
    # std_dev = 2*np.std(paths, axis=1)
    # plt.plot(t, std_dev, color='red', linewidth=2, label='2 Std Dev of Paths')
    # plt.plot(t, -std_dev, color='red', linewidth=2)

    deviation = np.sqrt(t[:])
    plt.fill_between(t, -deviation, deviation, color='red', alpha=0.2, label='±1 Std Dev')   
    deviation2 = 2*deviation
    plt.fill_between(t, -deviation2, deviation2, color='gray', alpha=0.5, label='±2 Std Dev')   

    plt.xlabel('Time')
    plt.ylabel('Position')
    plt.title(f'Brownian Motion - {num_paths} Paths')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

# Generate and plot 5 Brownian motion paths
plot_brownian_motion(num_paths=10, num_steps=50000, dt=0.1)
```
